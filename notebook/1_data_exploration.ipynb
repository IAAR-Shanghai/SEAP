```python
# notebooks/1_data_exploration.ipynb

import os
import pandas as pd
```

```python
# Define paths
root_data_dir = '../data/processed'
```

```python
# Load a sample dataset (can explore others based on task)
tasks_data_df = pd.read_parquet(os.path.join(root_data_dir, 'prompts.parquet'))
```

```python
# View the first few rows of the dataset
tasks_data_df[tasks_data_df['task_type']=='boolq'].head(5)
```

```python
# Check data types and missing values for each column
print(tasks_data_df.info())
```

```python
print(tasks_data_df[tasks_data_df['task_type']=='boolq'].iloc[0,7])
```

```python
# 获取所有唯一的 task_type
unique_tasks = tasks_data_df['task_type'].unique()

# 为每个 task_type 输出 iloc[0, 5]
for task in unique_tasks:
    subset = tasks_data_df[tasks_data_df['task_type'] == task]
    if len(subset) > 0:  # 确保至少有两行才有 iloc[1]
        value = subset.iloc[0, 5]
        print(f"{task:<15} → {value}")
    else:
        print(f"{task:<15} → [⚠️ 不足两行，跳过]")
```
