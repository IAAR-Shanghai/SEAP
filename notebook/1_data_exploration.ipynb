# Dataset Exploration and Analysis

This notebook explores the processed datasets to understand their structure, content, and characteristics across different tasks.

## Purpose
- Examine the structure and content of processed datasets
- Analyze data distributions across different tasks
- Validate data quality and completeness
- Explore prompt formats for different task types

## 1. Setup and Configuration

```python
import os
import pandas as pd

# Define data directory paths
root_data_dir = '../data/processed'
```

## 2. Load Dataset

```python
# Load the processed prompts dataset
tasks_data_df = pd.read_parquet(os.path.join(root_data_dir, 'prompts.parquet'))

# Display basic dataset information
print("Dataset Overview:")
print(tasks_data_df.info())
```

## 3. Data Analysis by Task Type

```python
# Example analysis for BoolQ task
print("Sample entries from BoolQ task:")
tasks_data_df[tasks_data_df['task_type']=='boolq'].head(5)
```

## 4. Prompt Format Analysis

```python
# Get all unique task types
unique_tasks = tasks_data_df['task_type'].unique()

# Display sample prompt format for each task type
print("Sample prompt format for each task type:")
for task in unique_tasks:
    subset = tasks_data_df[tasks_data_df['task_type'] == task]
    if len(subset) > 0:
        value = subset.iloc[0, 5]  # Sample prompt from first row
        print(f"{task:<15} → {value}")
    else:
        print(f"{task:<15} → [WARNING: No data available]")
```

## 5. Data Quality Analysis

Check for data completeness and quality issues.

```python
# Check for missing values
print("Missing values per column:")
print(tasks_data_df.isnull().sum())

# Display distribution of samples across tasks
print("\nSample distribution across tasks:")
print(tasks_data_df['task_type'].value_counts())
```
